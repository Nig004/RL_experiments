```python
#EXPERIMENT 01 : Grid-World + Q-Learning


import numpy as np, random

# 4x4 grid | Start:(0,0)  Goal:(3,3)  Hole:(1,1)
Q = np.zeros((4, 4, 4))          # Q[row][col][action]  4 actions: U D L R
GOAL, HOLE = (3,3), (1,1)

def step(s, a):
    r, c = s
    r, c = [r-1,r+1,r,r][a], [c,c,c-1,c+1][a]        # move
    r, c = max(0,min(3,r)), max(0,min(3,c))            # clamp to grid
    ns = (r,c)
    reward = 10 if ns==GOAL else -5 if ns==HOLE else -1
    return ns, reward, ns in [GOAL, HOLE]

# Q-Learning
for _ in range(500):
    s = (0,0)
    while True:
        r,c = s
        a = random.randint(0,3) if random.random()<0.2 else np.argmax(Q[r,c])
        ns, reward, done = step(s, a)
        nr,nc = ns
        # Q update: Q(s,a) += α[r + γ·maxQ(s',·) - Q(s,a)]
        Q[r,c,a] += 0.1*(reward + 0.9*np.max(Q[nr,nc]) - Q[r,c,a])
        s = ns
        if done: break

# Print policy
symbols = ['↑','↓','←','→']
print("Learned Policy:")
for r in range(4):
    print([' G ' if (r,c)==GOAL else ' H ' if (r,c)==HOLE
           else f' {symbols[np.argmax(Q[r,c])]} ' for c in range(4)])
```
