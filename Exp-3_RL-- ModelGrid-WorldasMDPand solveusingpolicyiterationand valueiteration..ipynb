```python
# Grid World MDP using Policy Iteration and Value Iteration

rows, cols = 3, 3
goal = (2, 2)
actions = ["U", "D", "L", "R"]
gamma = 0.9


def move(s, a):
    if s == goal:
        return s, 0

    x, y = s

    if a == "U":
        x -= 1
    elif a == "D":
        x += 1
    elif a == "L":
        y -= 1
    elif a == "R":
        y += 1

    if x < 0 or x >= rows or y < 0 or y >= cols:
        return s, -1

    ns = (x, y)

    if ns == goal:
        return ns, 10

    return ns, -1


states = [(i, j) for i in range(rows) for j in range(cols)]


# -------- Policy Iteration --------

policy = {s: "R" for s in states}
V = {s: 0 for s in states}

while True:
    # Policy Evaluation
    for _ in range(20):
        for s in states:
            if s != goal:
                ns, r = move(s, policy[s])
                V[s] = r + gamma * V[ns]

    # Policy Improvement
    stable = True
    for s in states:
        if s != goal:
            old = policy[s]
            policy[s] = max(actions, key=lambda a: move(s, a)[1] + gamma * V[move(s, a)[0]])
            if old != policy[s]:
                stable = False

    if stable:
        break

print("Policy Iteration Policy:")
for i in range(rows):
    for j in range(cols):
        print("T" if (i, j) == goal else policy[(i, j)], end=" ")
    print()


# -------- Value Iteration --------

V = {s: 0 for s in states}

for _ in range(30):
    for s in states:
        if s != goal:
            V[s] = max(move(s, a)[1] + gamma * V[move(s, a)[0]] for a in actions)

print("\nValue Iteration Policy:")
for i in range(rows):
    for j in range(cols):
        s = (i, j)
        if s == goal:
            print("T", end=" ")
        else:
            best = max(actions, key=lambda a: move(s, a)[1] + gamma * V[move(s, a)[0]])
            print(best, end=" ")
    print()
```
