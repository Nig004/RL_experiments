```python
# Policy Evaluation + Policy Improvement for a small MDP

gamma = 0.9
states = [0, 1, 2, 3]
actions = ["L", "R"]

# MDP: P[state][action] = (next_state, reward)
P = {
    0: {"L": (0, -1), "R": (1, 0)},
    1: {"L": (0, 0),  "R": (2, 0)},
    2: {"L": (1, 0),  "R": (3, 10)},
    3: {"L": (3, 0),  "R": (3, 0)}
}

policy = {s: "L" for s in states}
V = {s: 0 for s in states}

while True:
    # Policy Evaluation
    while True:
        delta = 0
        for s in states:
            if s == 3:
                continue

            old = V[s]
            a = policy[s]
            ns, r = P[s][a]
            V[s] = r + gamma * V[ns]

            delta = max(delta, abs(old - V[s]))

        if delta < 0.001:
            break

    # Policy Improvement
    stable = True
    for s in states:
        if s == 3:
            continue

        old = policy[s]

        policy[s] = max(
            actions,
            key=lambda a: P[s][a][1] + gamma * V[P[s][a][0]]
        )

        if old != policy[s]:
            stable = False

    if stable:
        break

print("Optimal Value Function:")
print(V)

print("\nOptimal Policy:")
for s in states:
    if s == 3:
        print("State", s, ": Terminal")
    else:
        print("State", s, ":", policy[s])
```
